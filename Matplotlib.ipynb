# Bar chart

* Bivarient analysis
* numerical vs categorical
* use case -> agregate analysis of groups

```python
#siple bar chart
children = [1,14,15,43,62]
colour=['red','blue','yellow','orange','black']
plt.bar(colour,children,color='green')
```

```python
#bar chart horizontal bars
#when numbers of categories increase on x axis

children = [1,14,15,43,62]
colour=['red','blue','yellow','orange','black']
plt.barh(colour,children,color='green')
```

```python
df=pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
#plot multiple bar for single quantity
plt.bar(np.arange(df.shape[0])- 0.2 ,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0])+ 0.2 ,df['2017'],width=0.2)
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
#when labels are long in length
children = [1,14,15,43,62]
colour=['red red red red','blue blue blue blue','yellow yellow yellow yellow','orange orange orange orange','black black black black']
plt.bar(colour,children)
plt.xticks(rotation='vertical')
plt.show()
```

# stack chart

```python
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2015')
plt.bar(df['batsman'],df['2017'],bottom=df['2015'] +df['2016'],label='2017')
plt.legend()
```

# pie chart

* univerient/bivarient analysis
* categorical vs numerical
* use case -> to find the contribution of each category on standard scale

```python
data=[12,34,56,78]
subjects=['english','science','urdu','math']
plt.pie(data,labels=subjects)
plt.show()
```

```python
#by using data
df=pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
```

```python
#give variety of colours
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['red','yellow','green','brown','black'])
plt.show()
```

```python
#explore perticular section
plt.pie(df['batsman_runs'],labels=df['batsman'],explode=[0.3,0,0,0.2,0,0])
```

```python
#give the shawdow to pie chart
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',shadow=True)
plt.show()
```
