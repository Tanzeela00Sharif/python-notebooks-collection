##split , combine , joining stratigy

```python
def num_of_genre(group):
  return group['Series_Title'].str.startswith('A').sum()

```

```python
genre.apply(num_of_genre)
```

```python
#give the each movia a ranking in each genre
def rank_movie(group):
  group['genre_rank']=group['IMDB_Rating'].rank(ascending= False)
  return group
```

```python
genre.apply(rank_movie)
```

##Normalized rating

```python
genre['IMDB_Rating'].agg(lambda x:x.max() -x.min()==0).sum()
```

```python
def normal(group):
  if  group['IMDB_Rating'].max() -group['IMDB_Rating'].min()==0:
    group['IMDB_Rating']==0
  else:
    group['ibmd_normalizaed']=(group['IMDB_Rating']- group['IMDB_Rating'].min()) /(group['IMDB_Rating'].max() -group['IMDB_Rating'].min())
    return group
```

```python
genre.apply(normal)
```

##**ipl dataset**

```python
ipl=pd.read_csv('/content/deliveries.csv')
```

```python
ipl.sample(5)
```

```python
#find the top 10 batsman in terms of runs
ipl.groupby('batsman')[ 'batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
#find  the number of batsman in term of max sixer
six=ipl[ipl['batsman_runs'] ==6]
six.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).index[0]
```

```python
ipl['over'].sample(5)
```

```python
#find batsman with most number of 4's and 5' in last five over
last_over=ipl[ipl['over']>15]
last_over=last_over[(last_over['batsman_runs'] ==4 )| (last_over['batsman_runs']==6)]
last_over.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head(1).index[0]
```

```python
#find virat kholi recrd against all teams
virat_batsman=ipl[ipl['batsman']=='V Kohli']
virat_batsman.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).reset_index()
```

```python
temp_df=ipl[ipl['batsman'] == "V Kohli"]
temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
#create a function that can return the highest score of any batsman
def highest_run(batsman):
  temp_df=ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```
