##way to create multiindx data


```python
1. #pd.MultiIndex.from_tuples
index_val=[('cse',2019),('cse',2020),('cse',2021),('cse',2022),('cse',2023),('cse',2024)]
a=pd.MultiIndex.from_tuples(index_val)
a.levels
```

```python
2. #pd.MultiIndex.from_product
MultiIndex=pd.MultiIndex.from_product([['cse','ece'],[2012,2013,2014,2015]])
```

```python
#creating series with multiindex
a=pd.Series([1,2,4,5,6,7,8,9],index=MultiIndex)
a
```

```python
#how to fetch items from such series
a[('ece',2012)]
a['cse']
```

```python
temp=a.unstack()
```

```python
temp.unstack()
```

##Melt and pivot dataframe

```python
data=pd.DataFrame({'cse':[120,23,12,13]})
data.melt()
```

```python
#wide data
D1=pd.DataFrame({'cse':[130],'ece':[123], 'mach':[50]})
D1.melt(var_name='subject',value_name='num_students')
```

```python
data=pd.DataFrame(
    {
      'branch':['cse','ece','mech','chemistry'],
      2012:[120,234,45,67],
      2014:[123,145,165,167],
      2015:[78,98,98,67]
    }
)
data.melt(id_vars=['branch'],var_name='year',value_name='num_student')
```
