```python
point_data=pd.DataFrame({
    '1st point':[(1,2),(3,4),(5,6),(6,7),(8,9)],
    '2nd point':[(1,2),(3,4),(5,6),(6,7),(8,9)]
})
point_data
```

```python
#apply function apply on row wise
def eulidean_distance(rows):
  point_A= rows['1st point']
  point_B= rows['2nd point']

  return ((point_A[0]-point_B[0])**2 +(point_A[1]-point_B[1])**2)**0.5


#call the function

point_data['difference']=point_data.apply(eulidean_distance,axis=1)

```
